```python
import pandas as pd
import numpy as np
np.random.seed(42)
names = ['Priya','Ravi','Anita','Suresh','Kavya','Mohammed','Deepa','Arjun','Sneha','Kiran']
data = {
    'Name':names,
    'Math':np.array([78,90,56,45,89,88,50,67,91,77]),
    'Science':np.array([55, 72, 89, 61, 77, 83, 94, 68, 79, 86]),
    'Science':np.array([91, 87, 73, 95, 82, 66, 88, 79, 84, 97]),
    'English':np.random.randint(40, 100, 10),
    'History':np.random.randint(40, 100, 10),
    'CS':np.random.randint(40, 100, 10),
    'Attendance': np.random.randint(55, 100, 10),
}
df = pd.DataFrame(data)
df.to_csv('student_raw.csv', index=False)
print('Raw data saved. Problems to find:', len(df))
print(df)
```

```python
df = pd.read_csv('student_raw.csv')
print('Shape:', df.shape)
print('\nMissing values:')
print(df.isnull().sum())
print('\nData types:')
print(df.dtypes)
print('\nBasic stats:')
print(df.describe())
df['Name'] = df['Name'].str.strip().str.title()
df.loc[df['Attendance'] > 100, 'Attendance'] = np.nan
subjects = ['Math', 'Science', 'English', 'History', 'CS']
for sub in subjects:
    df.loc[df[sub] < 0, sub] = np.nan
    df.loc[df[sub] > 100, sub] = np.nan
    for sub in subjects:
       df[sub] = df[sub].fillna(df[sub].median())
    df['Attendance'] = df['Attendance'].fillna(df['Attendance'].median())
print('\nClean data:')
print(df)
```

```python
df['Total']   = df[subjects].sum(axis=1)
df['Average'] = df[subjects].mean(axis=1).round(2)
def assign_grade(avg):
    if avg >= 90: return 'A+'
    elif avg >= 80: return 'A'
    elif avg >= 70: return 'B'
    elif avg >= 60: return 'C'
    else: return 'F'
df['Grade']  = df['Average'].apply(assign_grade)
df['Status'] = df.apply(
    lambda row: 'PASS' if row['Average'] >= 40 and row['Attendance'] >= 75
    else 'AT RISK' if row['Attendance'] < 75
    else 'FAIL', axis=1 )
print(f'Class Average:    {df["Average"].mean():.2f}')
print(f'Pass Rate:        {(df["Status"]=="PASS").mean()*100:.1f}%')
print(f'At-Risk Students: {(df["Status"]=="AT RISK").sum()}')
print(f'Top Scorer:       {df.loc[df["Average"].idxmax(), "Name"]}')
print(f'Weakest Subject:  {df[subjects].mean().idxmin()}')
df.to_csv('student_report.csv', index=False)
print('\nReport saved: student_report.csv')
```
